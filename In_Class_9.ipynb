Consider the `CarPrice_Assignment.csv` data file. This data is publicly available on the Kaggle website, and has information on cars (characteristics related to car dimensions, engine and more). The goal is to use car information to predict the price of the car. **In Python**, answer the following:

## 1. (4 points)

Using the `pandas` library, read the csv data file and create a data-frame called `car_price`.

```python
import pandas as pd 

car_price = pd.read_csv('CarPrice_Assignment.csv')
car_price.head()
```

## 2. (5 points)

Using the `wheelbase`, `enginesize`, `compressionratio`, `horsepower`, `peakrpm`, `citympg`, as the predictor variables, and `price` is the target variable. Define a 5-fold cross-validation strategy.

```python
from sklearn.model_selection import KFold

X = car_price[["wheelbase", "enginesize", "compressionratio", "horsepower", "peakrpm", "citympg"]].copy()
y = car_price["price"].copy()

skf = KFold(n_splits=5, shuffle=True, random_state=42)
```

## 3. (10 points)

Using the train dataset, build a support vector machine model (use `kernel = 'rbf'`). After that, use this model to predict on the test dataset. Report the 5-fold cross validation MSE of this model. Make sure to transform the input variables to 0-1 scale using the `MinMaxScaler` and `Pipeline`.

```python
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.svm import SVR
svm = make_pipeline(MinMaxScaler(),
                    SVR(kernel="rbf"))

cv_scores = cross_val_score(svm, X, y, cv=skf, scoring="neg_mean_squared_error", n_jobs=-1)
print(f"The mean cross-validated MSE is {-cv_scores.mean()}")
```

## 4. (10 points)

Using the train dataset, build a support vector machine model (use `kernel = 'poly'`). After that, use this model to predict on the test dataset. Report the 5-fold cross validation MSE of this model. Make sure to transform the input variables to 0-1 scale using the `MinMaxScaler` and `Pipeline`.

```python
svm = make_pipeline(MinMaxScaler(),
                    SVR(kernel="poly"))

cv_scores = cross_val_score(svm, X, y, cv=skf, scoring="neg_mean_squared_error", n_jobs=-1)
print(f"The mean cross-validated MSE is {-cv_scores.mean()}")
```

Using the results from parts (3) and (4), what model would you use to predict car prices? Explain.

I would use the second model SVR(kernel='poly) based on the results because there is a lower MSE which means less errors being made in the model.
